## Initial Observations

Based on the initial exploration:

1. The dataset contains 118614 records and 22 variables.

2. The target variable, `is_malicious`, is imbalanced, with
   94.62% normal records and 5.38% potentially malicious records.

3. The dataset contains 0 missing values and 0 duplicate records.

4. Some behavioural variables, particularly printing and file burning,
   appear to differ between normal and potentially malicious activity.

5. These observations will guide the feature-selection and
   machine-learning stages of the project.
